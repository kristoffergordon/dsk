```python
import pandas as pd

from dsk.ml import transform
from dsk.ml.model import MLModel
from dsk.ml.step import (
    BinColumn,
    OHE_Columns,
    ReplaceNaNs,
    SelectColumns,
    StandardizeScaler,
    ClipColumn,
    OrdinalEncodeColumns,
)
from dsk.ml.transform import Sequential
```

```python
# Create titanic data
Pclass = [1, 1, 3, 3, 1, 3, 3, 1, 3, 2]
Sex = ["m", "m", "f", "m", "f", "m", "m", "f", "m", "m"]
Age = [33.0, 28.0, 16.0, 44.0, 58.0, 20.0, 20.0, 18.0, 21.0, 54.0]
Survivied = [0, 0, 1, 1, 1, 0, 0, 1, 0, 0]

titanic_data = {"Pclass": Pclass, "Sex": Sex, "Age": Age, "Survivied": Survivied}

titanic_df = pd.DataFrame(titanic_data)
```

```python
bin_col = OrdinalEncodeColumns(cols=["Age", "Sex"])
```

```python
bin_col.fit_transform(titanic_df)
```
